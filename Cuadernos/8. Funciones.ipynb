<h2>Introducción</h2>
En Python, una función es un bloque de código diseñado para realizar una tarea específica, que puede ser reutilizado en distintas partes de un programa o incluso entre diferentes programas. Las funciones permiten organizar el código de manera más clara y estructurada, facilitando la resolución de problemas complejos al dividirlos en tareas más pequeñas y manejables.

El uso de funciones es fundamental para mejorar la eficiencia, el orden y la reutilización del código. Permiten encapsular operaciones repetitivas, manejar parámetros de entrada y obtener resultados, contribuyendo a un desarrollo más ordenado y menos propenso a errores.

Comprender la utilidad de las funciones es importante, ya que constituyen una herramienta clave para la construcción de programas modulables, escalables y fáciles de mantener.

Cree una función que divida la primera entrada por la segunda entrada:

```python
def divi(a, b):     #creamos la funcion llamada "divi", la cual dadas 2 entradas divide la primera entre la segunda
    return(a/b)
```

Utilice la función <code> con </code> para la siguiente pregunta.

```python
def con(a, b):  #se define la funcion como se indica
    return(a + b)
```

¿Se puede usar la función <code> con </code> que definimos antes para agregar números enteros o cadenas?

```python
#Si se puede usar para gregar numeros y para concatenar cadenas, ya que ambas entradas harán lo que pida la función, ejemplo

print(con(2, 5))
print(con("Hola ", "Juan"))
```

¿Se puede usar la función <code> con </code> que definimos antes para concentrar una lista o tupla?

```python
#Al igual que antes si se puede hacer, ya que "sumará" o "concatenará" ambas entradas, ejemplo

print(con(["Hola ", "Juan"], [18, "A"]))
```
