<h2>Introducción</h2>
En Python, los objetos son entidades que representan elementos del mundo real o conceptos abstractos dentro de un programa. Cada objeto combina información, mediante atributos, y comportamiento, mediante métodos, permitiendo modelar de manera organizada los elementos con los que se trabaja en una aplicación.

Comprender la importancia de los objetos es esencial, ya que proporciona una base sólida para la programación orientada a objetos, permitiendo desarrollar programas más estructurados, eficientes y fáciles de mantener.

El cuaderno enviado no cuenta con ejercicios enfocados en clases, asi que decidí crear una clase personas y empezar a experimentar con ella para practicar.

```python
class Persona(object):                  #definimos la clase Persona
    def __init__(self, name, age):      #Definimos su constructor que recibe un nombre y una edad
        self.name = name                #se asignan los valores de entrada
        self.age = age
    
    def birth(self):                    #definimos el metodo birth que añade un año a la edad actual
        self.age = self.age + 1
```

Crear una persona e imprimir sus atributos

```python
jugador = Persona("Juan", 18)                       #creamos a una persona

print("el nombre del jugador es", jugador.name)     #accedemos e imprimimos sus atributos
print("la edad del jugador es", jugador.age)
```

Crea un jugador nuevo, y cambiale el nombre

```python
jugador2 = Persona("Camilo", 19)                        #Creamos una segunda persona

print("Nombre antes del cambio:", jugador2.name)        #verificamos su nombre

jugador2.name = "Torres"                                #Cambiamos su nombre
print("nombre despues del cambio:", jugador2.name)      #Verificamos el cambio
```

El primer jugador cumplió años, usa el metodo "Birth" para cambiar su edad.

```python
print("Edad antes del cumpleaños:", jugador.age)        #Verificamos la edad

jugador.birth()                                         #Usamos el metodo para cambiar su edad
print("Edad despues del cumpleaños:", jugador.age)      #Verificamos la nueva edad
```
