<h2>Introducción</h2>
En Python, una tupla es una estructura de datos que permite almacenar una colección ordenada de elementos. Facilita la organización y el manejo de datos dentro de un programa. Las tuplas pueden contener diferentes tipos de datos.

El uso de tuplas es especialmente relevante cuando se requiere mantener la integridad de los datos, ya que esta estructura es inmutable. Esta propiedad contribuye a la seguridad y estabilidad de la información, evitando modificaciones accidentales durante la ejecución del programa.

Comprender el concepto y la utilidad de las tuplas resulta importante, ya que permite un manejo más estructurado de los datos y sienta las bases para el trabajo con listas y conjuntos.

Considérese la siguiente tupla:

```python
genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", \
                "R&B", "progressive rock", "disco") 
genres_tuple        #--Creación de tupla para el desarrollo del ejercicio
```

Encontrar la longitud de la tupla, <code>genres_tuple</code>:

```python
print(len(genres_tuple))  #--Imprimo la longitud de la tupla que creamos anteriormente
```

Accede al elemento en el índice 3:

```python
genres_tuple[3]     #-- se selecciona el indice 3 en la tupla
```

Utiliza el slicing para obtener los índices 3, 4 y 5:

```python
genres_tuple[3:6]  #ya que el segundo numero al que se le hace referencia no se muestra se tiene que poner 
                       #el siguiente
```

Encuentra los primeros dos elementos de la tupla <code>genres_tuple</code>:

```python
genres_tuple[0:2] #Con la misma logica anterior poniendo el indice del siguiente podemos encontrar los 2 primeros
```

Encuentra el primer índice de <code>"disco"</code>:

```python
genres_tuple.index("disco") #En tuplas el metodo index es el que nos permite encontrar lo que le indiquemos, en
                    #este caso es la palabra "disco" la cual esta en la posición 8 e indice 7.
```

Genera una lista ordenada desde la tupla <code>C_tuple=(-5, 1, -3)</code>:

```python
C_tuple = (-5, 1, -3)             #Creacion de tupla para el desarrollo del ejercicio
print(C_tuple)                    #impresión de la tupla antes del ordenamiento
sorted_tuple = sorted(C_tuple)    #el metodo sorted nos permite ordenar los elementos de una tupla de menor a mayor
print(sorted_tuple)               #impresión de la tupla despues de realizar el ordenamiento
```
