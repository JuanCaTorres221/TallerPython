<h2>Introducción</h2>
En Python, una lista es una estructura de datos que permite almacenar una colección ordenada de elementos. Las listas se utilizan para organizar y manejar conjuntos de datos de forma flexible, ya que pueden contener distintos tipos de información y adaptarse a las necesidades del programa durante su ejecución.

Comprender el uso y la importancia de las listas es importante, ya que constituyen una de las estructuras de datos más utiles y sirven como base para el desarrollo de soluciones más complejas y eficientes en el manejo de información.

Crea una lista <code>a_lit</code> con los siguientes elementos <code>1</code>, <code>hello</code>, <code>[1,2,3]</code> y <code>True</code>.

```python
a_list = [1, "hello", [1,2,3], True]   #Creacion de una lista con los valores asignados
```

Encuentra el valor almacenado en el índice 1 de <code>a_list</code>.

```python
print(a_list[1])   #en la lista se selecciona el indice 1
```

Obtén el elemento almacenado en el índice 1, 2 y 3 de <code>a_list</code>.

```python
print(a_list[1:4])      #se hace Slicing para imprimir los indices indicados
```

Concatena las siguientes listas <code>A = [1, 'a']</code> y <code>B = [2, 1, 'd']</code>:

```python
A = [1, 'a']        #Creación lista A
B = [2, 1, 'd']     #Creación lista B
print(A+B)          #Concatenamos los elementos de ambas listas y los imprimimos en pantalla
```
